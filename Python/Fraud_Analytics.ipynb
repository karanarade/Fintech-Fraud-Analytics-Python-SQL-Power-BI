```python
fraud_by_payment= df.groupby('Payment_Method')['Is_Fraud'].mean()*100
fraud_by_payment.sort_values(ascending=False)
```

```python
fraud_by_state= df.groupby('State')['Is_Fraud'].mean()*100
fraud_by_state.sort_values(ascending=False)
```

```python
fraud_by_type= df.groupby('Transaction_Type')['Is_Fraud'].mean()*100
fraud_by_type.sort_values(ascending=False)
```

```python
fraud_by_category= df.groupby('Merchant_Category')['Is_Fraud'].mean()*100
fraud_by_category.sort_values(ascending=False)
```

```python
fraud_by_Device_Type= df.groupby('Device_Type')['Is_Fraud'].mean()*100
fraud_by_Device_Type.sort_values(ascending=False)
```
